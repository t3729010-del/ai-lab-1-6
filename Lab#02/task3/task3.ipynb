```python
class ClassroomAgent:

    def __init__(self):
        self.previous_mode = None

    def act(self, temp, occupied):

        if not occupied:
            target = "ECO"
        elif temp > 26:
            target = "COOL"
        elif temp < 20:
            target = "WARM"
        else:
            target = "IDLE"

        if target == self.previous_mode:
            command = "NO COMMAND"
        else:
            command = "COMMAND"

        previous = self.previous_mode
        self.previous_mode = target

        return previous, target, command


agent = ClassroomAgent()

percepts = [(29, True), (29, True), (20, True),
            (26, True), (19, True), (29, False)]

for step, percept in enumerate(percepts, 1):

    previous, target, command = agent.act(percept[0], percept[1])

    print("Step:", step)
    print("Previous Mode:", previous)
    print("Percept:", percept)
    print("Target Mode:", target)
    print("Outcome:", command)
    print("-------------------")
```
