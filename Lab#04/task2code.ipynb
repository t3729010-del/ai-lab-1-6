```python
graph = {
    "S": [("A", 6), ("B", 9), ("C", 7)],
    "A": [("D", 8)],
    "B": [("D", 4), ("E", 2)],
    "C": [("E", 9)],
    "D": [("G", 1)],
    "E": [("G", 3)],
    "G": []
}

def ucs(start, goal):
    frontier = [(0, start, [start])]
    visited = []
    while frontier:
        frontier.sort(key=lambda x: x[0])
        cost, node, path = frontier.pop(0)
        if node in visited:
            continue
        visited.append(node)
        if node == goal:
            print("Visited:", visited)
            print("Path:", path)
            print("Total Cost:", cost)
            return
        for neighbour, weight in graph[node]:
            if neighbour not in visited:
                frontier.append((cost + weight, neighbour, path + [neighbour]))

ucs("S", "G")
```

```python
graph = {
    "S": [("A", 6), ("B", 9), ("C", 7)],
    "A": [("D", 8)],
    "B": [("D", 4), ("E", 2)],
    "C": [("E", 9)],
    "D": [("G", 1)],
    "E": [("G", 3)],
    "G": []
}
h = {"S": 0, "A": 1, "B": 2, "C": 3,
     "D": 6, "E": 4, "G": 0}

def astar(start, goal):
    frontier = [(0, 0, start, [start])]
    visited = []
    while frontier:
        frontier.sort()
        f, cost, node, path = frontier.pop(0)
        if node in visited:
            continue
        visited.append(node)
        if node == goal:
            print("Path:", path)
            print("Visited:", visited)
            print("Total Cost:", cost)
            return
        for neighbour, weight in graph[node]:
            if neighbour not in visited:
                g = cost + weight
                f = g + h[neighbour]
                frontier.append((f, g, neighbour, path + [neighbour]))

astar("S", "G")
```
