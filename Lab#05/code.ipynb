```python
import math

leaf_count = 0
def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):
    global leaf_count
    if curDepth == targetDepth:
        leaf_count += 1
        return scores[nodeIndex]
    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2, False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, False, scores, targetDepth)
        )
    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2, True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, True, scores, targetDepth)
        )
scores = [3, 5, 2, 9, 0, 7]
treeDepth = math.ceil(math.log2(len(scores)))
scores += [0] * (2**treeDepth - len(scores))

print("The optimal value is:", end=" ")
print(minimax(0, 0, True, scores, treeDepth))
print("Leaf Evaluation Count:", leaf_count)
```

```python
import math 

leaf_count = 0
def alphabeta(depth, index, maxTurn, scores, target, alpha, beta):
    global leaf_count
    if depth == target: 
        leaf_count += 1
        return scores[index] 
 
    if maxTurn: 
        value = float('-inf') 
        for i in range(2): 
            value = max(value, alphabeta(depth+1, index*2+i, False, scores, target, alpha, beta)) 
            alpha = max(alpha, value) 
            if alpha >= beta: 
                break 
        return value 
    else: 
        value = float('inf') 
        for i in range(2): 
            value = min(value, alphabeta(depth+1, index*2+i, True, scores, target, alpha, beta)) 
            beta = min(beta, value) 
            if alpha >= beta: 
                break 
        return value 
 
scores = [9,2,6,3,8,2,5,4,-1,0,2,1,9,6,5,3] 
depth = int(math.log2(len(scores))) 
 
value = alphabeta(0, 0, True, scores, depth, float('-inf'), float('inf'))

print("Optimal Value: ",value)
print("Leaf Evaluation Counter: ",leaf_count)
```
