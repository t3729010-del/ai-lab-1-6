```python
graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "D": ["G"],
    "C": ["E", "F"],
    "E": [],
    "F": [],
    "G": []
}

def dls(node, goal, limit, path):
    if node == goal:
        return "FOUND", path

    if limit == 0:
        return "CUTOFF", None

    cutoff = False

    for n in graph[node]:
        if n not in path:
            status, result = dls(n, goal, limit-1, path+[n])

            if status == "FOUND":
                return status, result
            if status == "CUTOFF":
                cutoff = True

    return ("CUTOFF", None) if cutoff else ("FAILURE", None)


for limit in range(1, 5):
    print(limit, dls("A", "G", limit, ["A"]))
```
