```python
graph = {
    "A": ["B", "C"],
    "B": ["D"],
    "D": ["G"],
    "C": ["E","F"],
    "E": [],
    "F": [],
    "G": []
}

q = []
visited = []

def bfs(start, goal):
    q.append((start, [start]))
    visited.append(start)

    while q:
        node, path = q.pop(0)

        if node == goal:
            return path, visited

        for n in graph[node]:
            if n not in visited:
                visited.append(n)
                q.append((n, path + [n]))
        print("Queue: ",[x[0] for x in q] ,"Visited: ",visited)

print(bfs("A", "G"))
```
